```python
def caesar_cipher(msg, shift, do='encrypt'): # define function for caesar cipher.
    # define the cipher to have a message, a shift amount, and to do an encryption or decryption
    result = ""
    if do == 'decrypt':
        shift = -shift
    for ch in msg: # for character in message
        if ch.isalpha(): # check if entry is a letter
            if ord(ch) >= 65 and ord(ch) <= 90: # if letter is between 65 and 90 it is lowercase
                s = chr((ord(ch) - 65 + shift) % 26 + 65)
            elif ord(ch) >= 97 and ord(ch) <= 122: #if letter is between 97 and 122 it is uppercase
                s = chr((ord(ch) - 97 + shift) % 26 + 97)
            result = result + s
        else:
            result = result + ch
    return result
    
def vigenere_cipher(msg, key, do='encrypt'): # define function for vigenere cipher
    # define the cipher to have a message, a string key, and to do an encryption or decryption
    result = ""
    j = 0
    key = key.upper() # convert all letters to uppercase
    for ch in msg: # for character in message
        if ch.isalpha(): # check if entry is a letter
            shift = ord(key[j%len(key)])-ord('A') # key shift
            if do == 'decrypt':
                shift = -shift
            if ord(ch) >= 65 and ord(ch) <= 90: # to check if lowercase
                s = chr((ord(ch)- 65 + shift) % 26 + 65)
            elif ord(ch) >= 97 and ord(ch) <= 122: # to check if uppercase
                s = chr((ord(ch)- 97 + shift) % 26 + 97)
            result = result + s
            j = j + 1
        else: 
            result = result + ch
    return result
    

user_choice_type = int(input("1: for Caesar cipher, 2: for Vigenere cipher")) # ask user to choose caesar or vigenere
if user_choice_type == 1: # if user picks caesar
    user_choice_by = int(input("Caesar cipher: 1- to encrypt a message, 2- to decrypt a message"))
    if user_choice_by == 1: # user wants to encrypt with caesar
        print("You have chosen to encrypt a message using a caesar cipher")
        msg = str(input("Add the message: "))
        shift = int(input("shift by: "))
        print(caesar_cipher(msg,shift,do='encrypt')) # call caesar_cipher function, and ask to encrypt
    else: # if user chooses to decrypt using caesar
        print("You have chosen to decrypt a message using a caesar cipher")
        msg = str(input("Add the message: ")) # user defined message
        shift = int(input("shift by: ")) # user defined numerical shift amount
        print(caesar_cipher(msg,shift,do='decrypt')) # call caesar_cipher function, and ask to decrypt
        
        
elif user_choice_type == 2: # if user chooses vigenere cipher
    user_choice_by = int(input("Vigenere cipher: 1- to encrypt a message, 2- to decrypt a message"))
    if user_choice_by == 1: # if user chooses to encrypt with vigenere cipher
        print("You have chosen to encrypt a message using a vigenere cipher")
        msg = str(input("Add the message: ")) # user inputted message
        key = str(input("key phrase: ")) # user inputted string/letter phrase
        print(vigenere_cipher(msg,key,do='encrypt')) # call vigenere_cipher function, and ask to encrypt
    else: # if user chooses to decrypt with vigenere cipher
        print("You have chosen to decrypt a message using a Vigenere cipher")
        msg = str(input("Add the message: "))
        key = str(input("key phrase: "))
        print(vigenere_cipher(msg,key,do='decrypt')) # call vigenere_cipher function, and ask to decrypt
```
